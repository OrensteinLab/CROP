```python
PATH_NORMAL_MESC = "../data/FORECasT_mESC.csv"
PATH_XCRISP_MESC = "../data/XCRISP_FORECasT_mESC.csv"
SAVE_PATH_MESC =  'output/xcrisp_vs_normal_correlation_line_plot_mESC.pdf'

PATH_NORMAL_TREX = "../data/FORECasT_K562_TREX2.csv"
PATH_XCRISP_TREX = "../data/XCRISP_FORECasT_TREX2.csv"
SAVE_PATH_TREX =  'output/xcrisp_vs_normal_correlation_line_plot_TREX.pdf'

PATH_NORMAL_HAP1 = "../data/FORECasT_HAP1.csv"
PATH_XCRISP_HAP1 = "../data/XCRISP_FORECasT_HAP1.csv"
SAVE_PATH_HAP1=  'output/xcrisp_vs_normal_correlation_line_plot_HAP1.pdf'

PATH_1 = PATH_NORMAL_MESC
PATH_2 = PATH_XCRISP_MESC
SAVE_PATH = SAVE_PATH_MESC

import pandas as pd
```

```python
df_normal = pd.read_csv(PATH_1)
df_xcrisp = pd.read_csv(PATH_2)
```

```python
# sort both datasets based on the 'sequence' column
df_normal = df_normal.sort_values(by='sequence').reset_index(drop=True)
df_xcrisp = df_xcrisp.sort_values(by='sequence').reset_index(drop=True)
```

```python
# only keep sequences in df_original where the sequence is also in df_xcrisp

indices_to_keep = df_normal['sequence'].isin(df_xcrisp['sequence'])
df_normal = df_normal[indices_to_keep].reset_index(drop=True)

print(len(df_normal))
print(len(df_xcrisp))
```

```python
# only keep sequencs in df_xcrisp if they are also in df_normal
indices_to_keep_xcrisp = df_xcrisp['sequence'].isin(df_normal['sequence'])
df_xcrisp = df_xcrisp[indices_to_keep_xcrisp].reset_index(drop=True)
print(len(df_xcrisp))
print(len(df_normal))
```

```python
# get all columns in df_normal that the column names are numerical including minus
numerical_cols = [col for col in df_normal.columns if col.replace('-', '').isdigit()]
print(numerical_cols)

# normalize all the values in each row (in numerical cols) to be 1
row_sums = df_normal[numerical_cols].sum(axis=1)
df_normal[numerical_cols] = df_normal[numerical_cols].div(row_sums, axis=0)

# get all the ones that are not 0 mod 3
non_mod3_cols = [col for col in numerical_cols if int(col) % 3 != 0]
print(non_mod3_cols)

# calculate the frameshift% column as the sum of values in those places
df_normal['fs'] = df_normal[non_mod3_cols].sum(axis=1)
```

```python
# get all columns in df_normal that the column names are numerical including minus
numerical_cols = [col for col in df_xcrisp.columns if col.replace('-', '').isdigit()]
print(numerical_cols)

# normalize all the values in each row (in numerical cols) to be 1
row_sums = df_xcrisp[numerical_cols].sum(axis=1)
df_xcrisp[numerical_cols] = df_xcrisp[numerical_cols].div(row_sums, axis=0)

# get all the ones that are not 0 mod 3
non_mod3_cols = [col for col in numerical_cols if int(col) % 3 != 0]
print(non_mod3_cols)

# calculate the frameshift% column as the sum of values in those places
df_xcrisp['fs'] = df_xcrisp[non_mod3_cols].sum(axis=1)
```

```python
# get correlation between the two frameshift% columns
correlation = df_normal['fs'].corr(df_xcrisp['fs'])
print(f"Correlation between frameshift% columns: {correlation}")
```

```python
# get correlation per all numerical columns in BOTH dataframes
common_numerical_cols = set(df_normal.columns).intersection(set(df_xcrisp.columns))
correlations = {}
for col in common_numerical_cols:
    if col.replace('-', '').isdigit():
        corr = df_normal[col].corr(df_xcrisp[col])
        correlations[col] = corr
# print correlations
for col, corr in correlations.items():
    print(f"Correlation for column {col}: {corr}")

# sort correlations by column name (numerically)
correlations = dict(sorted(correlations.items(), key=lambda item: int(item[0])))


# print the pearson correlation between numberical columns as a line plot
import matplotlib.pyplot as plt
# Set global font sizes for better readability
plt.rcParams.update({'font.size': 16}) # Base font size
plt.figure(figsize=(25, 8)) # Increased height slightly for larger labels

plt.plot(list(correlations.keys()), list(correlations.values()), marker='o', linewidth=2)

# Specific font size overrides
plt.xticks(rotation=90, fontsize=12) 
plt.yticks(fontsize=14)
plt.xlabel('$\Delta$length', fontsize=20, fontweight='bold')
plt.ylabel('Pearson correlation', fontsize=20, fontweight='bold')

plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig(SAVE_PATH, dpi=300)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Global font settings for publication-quality plots
plt.rcParams.update({'font.size': 18})

# Define paths (Update these to your local environment)
PATH_NORMAL_MESC = "../data/FORECasT_mESC.csv"
PATH_XCRISP_MESC = "../data/XCRISP_FORECasT_mESC.csv"
PATH_NORMAL_TREX = "../data/FORECasT_K562_TREX2.csv"
PATH_XCRISP_TREX = "../data/XCRISP_FORECasT_TREX2.csv"
PATH_NORMAL_HAP1 = "../data/FORECasT_HAP1.csv"
PATH_XCRISP_HAP1 = "../data/XCRISP_FORECasT_HAP1.csv"

def draw_side_by_side_profiles(file_normal, file_xcrisp, name, save_path, normalize=True):
    """
    Generates side-by-side comparison plots of mutation profiles.
    Normalizes by sum (Probability Density) and supports optional log scaling.
    """
    # Load data
    df_normal = pd.read_csv(file_normal)
    df_xcrisp = pd.read_csv(file_xcrisp)

    # Identify and sort numerical columns (bins) by integer value
    cols_n = sorted([c for c in df_normal.columns if c.strip('-').isdigit()], key=int)
    cols_x = sorted([c for c in df_xcrisp.columns if c.strip('-').isdigit()], key=int)

    # Calculate mean frequency and fill NaNs with zero
    profile_n = df_normal[cols_n].fillna(0).mean()
    profile_x = df_xcrisp[cols_x].fillna(0).mean()

    # Convert indices to integers for proper reindexing
    profile_n.index = profile_n.index.map(int)
    profile_x.index = profile_x.index.map(int)

    # Create a continuous range of bins to avoid gaps in the line plot
    all_bins = range(min(min(profile_n.index), min(profile_x.index)), 
                     max(max(profile_n.index), max(profile_x.index)) + 1)
    
    profile_n = profile_n.reindex(all_bins, fill_value=0)
    profile_x = profile_x.reindex(all_bins, fill_value=0)

    # Normalization by SUM (Total probability = 1)
    if normalize:
        sum_n = profile_n.sum()
        sum_x = profile_x.sum()
        if sum_n > 0: profile_n = profile_n / sum_n
        if sum_x > 0: profile_x = profile_x / sum_x

    # Create figure with 1 row and 2 columns
    fig, axes = plt.subplots(1, 2, figsize=(22, 8), sharey=True)
    
    colors = ['#1f77b4', '#ff7f0e']
    profiles = [profile_n, profile_x]
    titles = [f'FORECasT', f'X-CRISP']

    for i in range(2):
        axes[i].plot(profiles[i].index, profiles[i].values, color=colors[i], linewidth=2, marker='.', markersize=4)
        axes[i].set_title(titles[i] + ('' if normalize else ''), fontsize=22, pad=20)
        axes[i].set_xlabel(r'$\Delta$length', fontsize=20)
        axes[i].grid(True, which="both", linestyle='--', alpha=0.4)
        

        curr_max = profiles[i].max()
        axes[i].set_ylim(-0.01 * curr_max, curr_max * 1.1)

        # Restrict view to -50 to 50 as requested
        axes[i].set_xlim(-50, 50)

    axes[0].set_ylabel('Probability ', fontsize=20)

    plt.tight_layout()
    plt.savefig(save_path, bbox_inches='tight')
    plt.show()
    plt.close()

# Execution loop
datasets = [
    (PATH_NORMAL_HAP1, PATH_XCRISP_HAP1, "HAP1", "output/profile_HAP1.pdf"),
    (PATH_NORMAL_TREX, PATH_XCRISP_TREX, "TREX", "output/profile_TREX.pdf"),
    (PATH_NORMAL_MESC, PATH_XCRISP_MESC, "mESC", "output/profile_mESC.pdf")
]

for n, x, name, path in datasets:
    draw_side_by_side_profiles(n, x, name, path)
```
