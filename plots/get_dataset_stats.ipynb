# General

```python
path = "../data"

# get all csv files in the data path
import pandas as pd
import glob
csv_files = glob.glob(f"{path}/*.csv")
print(f"Found {len(csv_files)} CSV files.")

# print all of them
for file in csv_files:
    print(file)
```

```python
# remove SPROUT_T_CROTON_VERSION and all that have XCRISP_FORECasT or Lindel in the filename
filtered_files = [f for f in csv_files if  "XCRISP_FORECasT" not in f and "Lindel" not in f]
print(f"Filtered down to {len(filtered_files)} CSV files.")
```

```python
# get n_rows in each csv file and print and also sum them up
total_rows = 0
for file in filtered_files:
    df = pd.read_csv(file)
    # print number of rows before filtering
    print(f"{file}: {df.shape[0]} ", end='; ')
    # remove all rows with sum of 0 in the numerical columns (assuming first column is non-numeric)
    num_cols = []
    for col in df.columns:
        try:
            float(col)
            num_cols.append(col)
        except ValueError:
            continue
    df = df[(df[num_cols].sum(axis=1) != 0)]
    n_rows = df.shape[0]
    total_rows += n_rows
    print(f"{n_rows} rows")

print(f"Total rows across all filtered CSV files: {total_rows}")
```

```python
# for each dataset, remove all columns that are filled with 0s, then report the numerical columns remaining (as a single line print)

for file in filtered_files:
    df = pd.read_csv(file)
    # remove columns that are all 0s
    df = df.loc[:, (df != 0).any(axis=0)]
    # get all columns that can be mapped to numbers
    num_cols = []
    for col in df.columns:
        try:
            float(col)
            num_cols.append(col)
        except ValueError:
            continue

    # transform num_cols into intgers (they are all integers)
    num_cols = [int(col) for col in num_cols]
    min_col = min(num_cols)
    max_col = max(num_cols)
    print(f"{file}: {min_col} to {max_col} ({len(num_cols)} columns)")
```

```python
# for each dataset get the min and max sequence length
for file in filtered_files:
    df = pd.read_csv(file)
    seq_lengths = df['sequence'].apply(len)
    min_len = seq_lengths.min()
    max_len = seq_lengths.max()
    print(f"{file}: min length = {min_len}, max length = {max_len}")
```

```python
# for each file, get the min and max PAM position
for file in filtered_files:
    df = pd.read_csv(file)
    pam_positions = df['PAM position']
    min_pam = pam_positions.min()
    max_pam = pam_positions.max()
    print(f"{file}: min PAM position = {min_pam}, max PAM position = {max_pam}")
    # print all positions as well
    print(f"Positions: {sorted(pam_positions.unique())}")
```

# FORECasT

```python
path = "../data"

# get all csv files in the data path
import pandas as pd
import glob
csv_files = glob.glob(f"{path}/*.csv")


# remove SPROUT_T_CROTON_VERSION and all that have XCRISP_FORECasT or Lindel in the filename
filtered_files = [f for f in csv_files if  "XCRISP_FORECasT" not in f]
#print(f"Filtered down to {len(filtered_files)} CSV files.")


# load all forecast datasets
forecast_files = [f for f in filtered_files if "FORECasT" in f]
print(f"Found {len(forecast_files)} FORECasT CSV files.")
```

```python
all_targets = set()
for file in forecast_files:
    df = pd.read_csv(file)
    targets = set(df['sequence'].unique())
    all_targets.update(targets)

print(f"Total unique target sequences across all FORECasT datasets: {len(all_targets)}")
```
